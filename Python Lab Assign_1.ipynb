```python
n = int(input("Enter a number: "))

for i in range(1, 11):
    print(n, "x", i, "=", n * i)
```

```python
n = int(input("Enter number of rows: "))

for i in range(1, n + 1):
    print("*" * i)
```

```python
n = int(input("Enter number of rows: "))

for i in range(n, 0, -1):
    print("*" * i)
```

```python
n = int(input("Enter number of rows: "))

for i in range(1, n + 1):
    print(" " * (n - i) + "*" * (2 * i - 1))
```

```python
n = int(input("Enter number of terms: "))

a = 0
b = 1

for i in range(n):
    print(a, end=" ")
    c = a + b
    a = b
    b = c
```
