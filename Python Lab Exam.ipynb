```python
p = input("Enter a paragraph: ")

p = p.lower()
p = p.replace(",", "")
p = p.replace(".", "")
p = p.replace("!", "")

words = p.split()
unique = set(words)

print("Total words:", len(words))
print("Unique words:", len(unique))

repeated = set()

for w in words:
    if words.count(w) > 1:
        repeated.add(w)

print("Repeated words:", repeated)

print("Words in alphabetical order:")
print(sorted(unique))
```
